### Ejercicio 8 (Aplicando a proyectos):
Para diversificar y atraer nuevos clientes, una lanchonete creó un ítem misterioso en su menú llamado "ensalada de frutas sorpresa". En este ítem, se eligen aleatoriamente 3 frutas de una lista de 12 para componer la ensalada de frutas del cliente. Crea el código que realice esta selección aleatoria según la lista dada:
`frutas = ["manzana", "banana", "uva", "pera", "mango", "coco", "sandia", "fresa", "naranja", "maracuya", "kiwi", "cereza"]`

```python
import random

frutas = ["manzana", "banana", "uva", "pera", "mango", "coco", "sandia", "fresa", "naranja", "maracuya", "kiwi", "cereza"]

# Usa random.sample() para seleccionar 3 elementos únicos de la lista 'frutas'.
# 'random.sample' garantiza que no haya repeticiones en la selección.
frutas_seleccionadas = random.sample(frutas, 3)

print("Las frutas seleccionadas son:", frutas_seleccionadas)
```

```python
# Este código utiliza numpy.random.choice para seleccionar frutas con probabilidades personalizadas.
# Esto es útil para simular un 'control de inventario' o darle preferencia a ciertas frutas.
import numpy as np

frutas = ["manzana", "banana", "uva", "pera", "mango", "coco", "sandia", "fresa", "naranja", "maracuya", "kiwi", "cereza"]

# Define las probabilidades para cada fruta. La suma de estas probabilidades no tiene que ser 1,
# ya que se normalizarán.
probabilidades = [
    0.02,  # Manzana (baja chance, para conservarla)
    0.15,  # Banana (alta chance, para gastarla)
    0.083, # Uva (cerca del promedio)
    0.083, # Pera
    0.083, # Mango
    0.083, # Coco
    0.083, # Sandia
    0.083, # Fresa
    0.083, # Naranja
    0.083, # Maracuya
    0.083, # Kiwi
    0.043  # Cereza (media-baja chance)
]

# Calcula la suma total de las probabilidades dadas.
suma_total = sum(probabilidades)

# Normaliza las probabilidades para que su suma sea 1.
# Esto es un requisito para el parámetro 'p' de np.random.choice.
probabilidades_normalizadas = [prob / suma_total for prob in probabilidades]

# Selecciona 3 frutas de la lista, sin reemplazo (replace=False),
# y utilizando las probabilidades normalizadas ('p').
frutas_seleccionadas = np.random.choice(frutas, 3, replace=False, p=probabilidades_normalizadas)

print("Las frutas seleccionadas son:", frutas_seleccionadas)
```

Has recibido un desafío para calcular la raíz cuadrada de una lista de números, identificando cuáles resultan en un número entero. La lista es la siguiente:

numeros = [2, 8, 15, 23, 91, 112, 256]

```python
import numpy as np

numeros = [2, 8, 15, 23, 91, 112, 256]

lista_enteros = []

# Itera sobre cada número en la lista.
for numero in numeros:
    # Calcula la raíz cuadrada del número usando np.sqrt().
    raiz_cuadrada = np.sqrt(numero)
    # Compara la raíz cuadrada (flotante) con su versión entera.
    # Si son iguales, significa que la raíz cuadrada es un número entero.
    if raiz_cuadrada == int(raiz_cuadrada):
        lista_enteros.append(numero)

print("Los números que tienen como raíz cuadrada un número entero son:", lista_enteros)
```

```python
import numpy as np

numeros = [2, 8, 15, 23, 91, 112, 256]

# 1. Convertir la lista Python a un array de NumPy.
array_numeros = np.array(numeros)

# 2. Calcular las raíces cuadradas de todos los elementos del array de una vez (operación vectorizada).
raices = np.sqrt(array_numeros)

# 3. Determinar si cada raíz cuadrada es un número entero.
# np.floor(raices) redondea hacia abajo al entero más cercano.
# Si la raíz cuadrada es igual a su piso, entonces es un entero.
es_entero = np.floor(raices) == raices

# 4. Filtrar el array original para obtener solo los números cuyas raíces cuadradas son enteras.
numeros_filtrados = array_numeros[es_entero]

print("El número que tiene como raíz cuadrada un número entero es el:", numeros_filtrados)
```

### Ejercicio 10:
Haz un programa para una tienda que vende césped para jardines. Esta tienda trabaja con jardines circulares y el precio del metro cuadrado de césped es de R$ 25,00. Pide a la persona usuaria el radio del área circular y devuelve el valor en reales de cuánto tendrá que pagar.

```python
import math

precio_metro_cuadrado = 25.00

# Solicita al usuario que ingrese el radio del área circular y lo convierte a tipo flotante.
radio = float(input("Ingrese el radio del área circular que desea cubrir: "))

# Calcula el área del círculo usando la fórmula A = π * radio^2.
# math.pi proporciona el valor de PI y '** 2' calcula el cuadrado del radio.
area = math.pi * radio ** 2

# Calcula el costo total multiplicando el área por el precio por metro cuadrado.
costo_total = area * precio_metro_cuadrado

# Imprime el resultado formateado a dos decimales.
print(f"El costo total para cubrir el área circular de radio {radio} es de R$ {costo_total:.2f}")
```

```python
import numpy as np

precio_metro_cuadrado = 25.00

# Solicita al usuario el radio del área circular.
radio = float(input("Ingrese el radio del área circular que desea cubrir: "))

# Calcula el área del círculo usando np.pi para el valor de PI.
# np.pi es la constante PI de la biblioteca NumPy.
area = np.pi * radio ** 2

# Calcula el costo total.
costo_total = area * precio_metro_cuadrado

# Imprime el costo total formateado a dos decimales.
print(f"El costo total para cubrir el área circular de radio {radio} es de R$ {costo_total:.2f}")
```

```python
import math

precio_metro_cuadrado = 25.00
# Solicita al usuario el radio del área circular.
radio = float(input("Ingrese el radio del área circular en metros: "))
# Calcula el área utilizando math.pi y math.pow para la potencia.
area = math.pi * math.pow(radio, 2)
# Calcula el costo total.
costo_total = precio_metro_cuadrado * area

# Imprime el costo total redondeado a dos decimales.
print(f"El costo total es de R$ {round(costo_total,2)}")
```
