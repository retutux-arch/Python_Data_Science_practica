### Ejercicio 4

Crea una función que reciba dos listas como parámetros y agrupe los elementos uno a uno de las listas, formando una lista de tuplas de 3 elementos. El primer y segundo elemento de la tupla son los valores en la posición i de las listas y el tercer elemento es la suma de los valores en la posición i de las listas.

La función debe incluir un manejo de error indicando el tipo de error generado y devolver como resultado la lista de tuplas. Si las listas enviadas como parámetro tienen tamaños diferentes, la función debe devolver un IndexError con la frase: 'La cantidad de elementos en cada lista es diferente.'.

Datos para probar la función:

Valores sin error:

`lista1 = [4, 6, 7, 9, 10]`
`lista2 = [-4, 6, 8, 7, 9]`

Listas con tamaños diferentes:

`lista1 = [4, 6, 7, 9, 10, 4]`
`lista2 = [-4, 6, 8, 7, 9]`

Listas con valores incoherentes:

`lista1 = [4, 6, 7, 9, 'A']`
`lista2 = [-4, 'E', 8, 7, 9]`

```python
def agrupar_listas(lista1, lista2):
    try:
        if len(lista1) != len(lista2):
            raise IndexError("La cantidad de elementos en cada lista es diferente.")

        lista_tuplas = []
        for i in range (len(lista1)):
            suma = lista1[i] + lista2[i]
            tupla = (lista1[i], lista2[i], suma)
            lista_tuplas.append(tupla)
        return lista_tuplas
    except IndexError as e:
        print(f"Error: {e}")
        return []
    except TypeError as e:
        print(f"Error: {e}")
        return []


lista1 = [4, 6, 7, 9, 10]
lista2 = [-4, 6, 8, 7, 9]
lista_tratada = agrupar_listas(lista1, lista2)
print(lista_tratada)

lista1 = [4, 6, 7, 9, 10, 4]
lista2 = [-4, 6, 8, 7, 9]
lista_tratada = agrupar_listas(lista1, lista2)
print(lista_tratada)

lista1 = [4, 6, 7, 9, 'A']
lista2 = [-4, 'E', 8, 7, 9]
lista_tratada = agrupar_listas(lista1, lista2)
print(lista_tratada)

```

```python
#codigo instructor

def agrupar_listas(lista1, lista2):
    try:
        if len(lista1) != len(lista2):
            raise IndexError("La cantidad de elementos en cada lista es diferente.")

        lista_resultado = [(lista1[i], lista2[i], lista1[i] + lista2[i]) for i in range(len(lista1))]
        return lista_resultado
    except TypeError as e:
        print(f"Error: {e}")
    except IndexError as e:
        print(f"Error: {e}")

# Ejemplos de uso:
try:
    resultado_sin_error = agrupar_listas([4, 6, 7, 9, 10], [-4, 6, 8, 7, 9])
    print(resultado_sin_error)

    resultado_tamanos_diferentes = agrupar_listas([4, 6, 7, 9, 10, 4], [-4, 6, 8, 7, 9])
    print(resultado_tamanos_diferentes)

    resultado_valores_incoherentes = agrupar_listas([4, 6, 7, 9, 'A'], [-4, 'E', 8, 7, 9])
    print(resultado_valores_incoherentes)
except Exception as e:
    print(f"Error general: {e}")
```

### Ejercicio 5

Como desafío, se te ha asignado la tarea de desarrollar un código que contabiliza las puntuaciones de estudiantes de una institución educativa de acuerdo con sus respuestas en una prueba. Este código debe ser probado para un ejemplo de 3 estudiantes con una lista de listas en la que cada lista tiene las respuestas de 5 preguntas objetivas de cada estudiante. Cada pregunta vale un punto y las alternativas posibles son A, B, C o D.

Si alguna alternativa en una de las pruebas no está entre las alternativas posibles, debes lanzar un ValueError con el mensaje "La alternativa [alternativa] no es una opción de alternativa válida". El cálculo de las 3 notas solo se realizará mediante las entradas con las alternativas A, B, C o D en todas las pruebas. Si no se lanza la excepción, se mostrará una lista con las notas en cada prueba.

Datos para la prueba del código:

Respuestas de la prueba:

`respuestas = ['D', 'A', 'B', 'C', 'A']`

A continuación, hay 2 listas de listas que puedes usar como prueba:

Notas sin excepción:

`tests_sin_ex = [['D', 'A', 'B', 'C', 'A'], ['C', 'A', 'A', 'C', 'A'], ['D', 'B', 'A', 'C', 'A']]`

Notas con excepción:

`tests_con_ex = [['D', 'A', 'B', 'C', 'A'], ['C', 'A', 'A', 'E', 'A'], ['D', 'B', 'A', 'C', 'A']]`

```python
respuestas = ['D', 'A', 'B', 'C', 'A']

def calcular_notas(tests, respuestas_correctas):
    notas = []
    alternativas_validas = ['A', 'B', 'C', 'D']

    try:
        for test_estudiante in tests:
            nota_actual = 0
            for i, respuesta_estudiante in enumerate(test_estudiante):
                # Validar que la alternativa sea válida
                if respuesta_estudiante not in alternativas_validas:
                    raise ValueError(f"La alternativa {respuesta_estudiante} no es una opción de alternativa válida")

                # Comparar con las respuestas correctas
                if respuesta_estudiante == respuestas_correctas[i]:
                    nota_actual += 1
            notas.append(nota_actual)
        return notas
    except ValueError as e:
        print(f"Error: {e}")
        return []
# Pruebas con excepción
tests_con_ex = [['D', 'A', 'B', 'C', 'A'], ['C', 'A', 'A', 'E', 'A'], ['D', 'B', 'A', 'C', 'A']]
notas_calculadas_con_ex = calcular_notas(tests_con_ex, respuestas)
print(f"Notas con excepción: {notas_calculadas_con_ex}")

# Pruebas sin excepción
tests_sin_ex = [['D', 'A', 'B', 'C', 'A'], ['C', 'A', 'A', 'C', 'A'], ['D', 'B', 'A', 'C', 'A']]
notas_calculadas_sin_ex = calcular_notas(tests_sin_ex, respuestas)
print(f"Notas sin excepción: {notas_calculadas_sin_ex}")
```

```python
#codigo instructor

respuesta = ['D', 'A', 'B', 'C', 'A']
tests_sin_ex = [['D', 'A', 'B', 'C', 'A'], ['C', 'A', 'A', 'C', 'A'], ['D', 'B', 'A', 'C', 'A']]
tests_con_ex = [['D', 'A', 'B', 'C', 'A'], ['C', 'A', 'A', 'E', 'A'], ['D', 'B', 'A', 'C', 'A']]

def corrector(tests: list):
  puntuaciones = []
  try:
    for test in tests:
      nota = 0
      for i in range(len(test)):
        if test[i] not in ['A', 'B', 'C', 'D']:
          raise ValueError(f'La alternativa {test[i]} no es una opción de alternativa válida')
        elif test[i] == respuesta[i]:
          nota += 1
      puntuaciones.append(nota)
  except Exception as e:
    print(e)
  else:
    return puntuaciones

corrector(tests_con_ex)
corrector(tests_sin_ex)
```
