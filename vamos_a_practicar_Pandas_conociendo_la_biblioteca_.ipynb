## **3. Aplica un filtro que seleccione solo a los estudiantes que fueron aprobados.**



```python
datos_alumnos
```

```python
# El filtro esta correcto, pero la columna "Aprobados" no es consistente
# ya que todos los valores son objetos en lugar de booleanos
df_alumnos_aprobados = datos_alumnos.query('Aprobado == True')
df_alumnos_aprobados
```

```python
datos_alumnos.info()
```

```python
# con el metodo ".eplace" cambiamos Verdadero por True
datos_alumnos['Aprobado'] = datos_alumnos['Aprobado'].replace('Verdadero', True)
#
#datos_alumnos.loc[datos_alumnos['Aprobado'] == 'Verdadero', 'Aprobado'] = True
```

```python
# Los datos de la columna "Aprobados son objetos no bool, por lo cual reguilarizamos"
# Aseguramos que la columna 'Aprobado' sea de tipo string para un mapeo consistente
datos_alumnos['Aprobado'] = datos_alumnos['Aprobado'].astype(str)

# Definimos un diccionario de mapeo para convertir strings a booleanos
# Esto manejará 'Verdadero', 'True' y 'False' correctamente
conversion_map = {
    'Verdadero': True,
    'True': True,
    'False': False
}

# Aplicamos el mapeo a la columna 'Aprobado'
datos_alumnos['Aprobado'] = datos_alumnos['Aprobado'].map(conversion_map)

# Ahora, convertimos explícitamente la columna a tipo booleano.
# Esto es importante para asegurar que Pandas la reconozca como 'bool' y no 'object'.
datos_alumnos['Aprobado'] = datos_alumnos['Aprobado'].astype(bool)

# Filtramos solo a los estudiantes que fueron aprobados
df_alumnos_aprobados = datos_alumnos.loc[datos_alumnos['Aprobado'] == True]

# Mostramos el DataFrame filtrado y el tipo de dato de la columna 'Aprobado' para verificar
display(df_alumnos_aprobados)
display(datos_alumnos['Aprobado'].dtypes)
```

```python
# object a bool
datos_alumnos.info()
```

```python
# Filtro operando
df_alumnos_aprobados
```

```python
# con "reset._index" reemplazmos los indices numericos para mejor
# legibilidad y manejo futuro
df_alumnos_aprobados = df_alumnos_aprobados.reset_index(drop=True)
df_alumnos_aprobados
```

#  **Practica biblitecas Pandas 2.2**

## **1. Los estudiantes participaron en una actividad extracurricular y ganaron puntos extras. Estos puntos extras corresponden al 40% de su nota actual. Por lo tanto, crea una columna llamada "Puntos_extras" que contenga los puntos extras de cada estudiante, es decir, el 40% de su nota actual.**

```python
datos_alumnos['Puntos_extras'] = datos_alumnos['Nota'] * 0.4
datos_alumnos
```

## **2. Crea otra columna llamada "Notas_finales" que contenga las notas de cada estudiante sumadas con los puntos extras.**

```python
datos_alumnos['Notas_finales'] = datos_alumnos['Nota'] + datos_alumnos['Puntos_extras']
datos_alumnos
```

## **3. Dado que hubo una puntuación extra, algunos estudiantes que no habían sido aprobados antes pueden haber sido aprobados ahora. En función de esto, crea una columna llamada "Aprobado_final" con los siguientes valores:**


* ***True: si el estudiante está aprobado (la nota final debe ser mayor o igual a 7.0).***

* ***False: si el estudiante está reprobado (la nota final debe ser menor que 7.0).***






```python
datos_alumnos['Aprobado_final'] = datos_alumnos['Notas_finales'].apply(lambda x: True if x >= 7.0 else False)
datos_alumnos
```

## **4. Realiza una selección y verifica qué estudiantes no habían sido aprobados anteriormente, pero ahora fueron aprobados después de sumar los puntos extras.**

```python
alumnos_reprobados = datos_alumnos.query('Aprobado == False and Aprobado_final == True')
alumnos_reprobados
```
